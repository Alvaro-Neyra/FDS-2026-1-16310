Al realizar, el cálculo de tablas de frecuencias absolutas y relativas de todas las variables permitió identificar severas inconsistencias estructurales y de formato en las variables del dataset original, causadas por errores de importación y el uso incorrecto de comillas en la fuente de YouTube:

* **`video_id`:** Presenta registros corruptos con valores `#NAME?` (1.33% del total) y caracteres de salto de línea (`\n`). Asimismo, evidencia un desplazamiento físico de columnas que provocó el almacenamiento erróneo de textos de descripción en este campo.

* **`views`:** Al ser una variable cuantitativa continua, sus porcentajes de repetición individual son marginales (0.01%). El problema crítico radica en que el sistema leyó el 96.9% de sus celdas como cadenas de texto (`str`) y el resto como flotantes debido a la presencia de nulos (`NaN`), bloqueando los cálculos estadísticos descriptivos directos.
* **`likes`, `dislikes` y `comment_count`:** Comparten la misma anomalía de tipos mezclados (`str` y `float`). Adicionalmente, acumulan frecuencias en el valor `0.0` (asociadas a videos con el panel de interacciones o comentarios desactivados desde el origen de la publicacion del video).

Los hallazgos confirman la presencia de ruido y tipos de datos incorrectos, justificando técnicamente la necesidad de un preprocesamiento exhaustivo para corregir y normalizar la estructura de las variables antes de proceder al modelado.

**Verificar valores negativos:**

```python
# Seleccionamos todas las columnas que son numéricas
numericas = data.select_dtypes(include=['float64', 'int'])

print("Análisis de variables numéricas en el dataset crudo (solo float64 e int):")
for col in numericas.columns:
    serie_convertida_a_numerica = pd.to_numeric(numericas[col], errors='coerce')

    negativos = (serie_convertida_a_numerica < 0).sum()

    nulos_o_no_numericos = serie_convertida_a_numerica.isna().sum()

    print(f"Columna: {col:15} | Negativos: {negativos:4} | Valores no-numéricos o nulos: {nulos_o_no_numericos:5}")
```

Al cruzar los resultados de las tablas de frecuencia con el diagnóstico de valores negativos, se identificó un patrón crítico: las variables numéricas (`views`, `likes`, `dislikes` y `comment_count`) registran exactamente la misma cantidad de valores nulos o no numéricos (1,181 registros). Este hallazgo confirma que no estamos ante omisiones aleatorias de datos, sino ante un error de formateo estructural por desplazamiento físico de columnas, originado por la presencia de saltos de línea físicos dentro del campo de texto libre description.

**Tratar las variables `video_id` con un correcto formato**

En la etapa de inspeccion de datos, hemos descubierto que hay valores para la variable `video_id` que no son correctas. Por lo que antes de eliminar variables que no necesitamos lo mejor seria tratar primero los video_ids, luego hallar los duplicados (agrupados por trending_date y video_id, ademas de las filas que estan exactamente duplicadas, aunque podrian tener el mismo id).



**Correccion de IDs**

Se identificaron 511 registros con video_id igual a `#NAME?` y 7 registros con `#VALUE!`, errores típicos que genera Excel cuando un identificador de video comienza con caracteres especiales (- `o` `=`). Dado que estas filas contienen datos métricos y geográficos reales completamente válidos, no deben ser eliminadas. Para evitar que el que se detecten como falsos duplicados, verificaremos si existen videos con data igual para reemplazar su id por el existente en caso contrario, generaremos de manera programática un identificador temporal único aplicando un algoritmo hash sobre la combinación de las columnas `title` y `channel_title`.

```python
import hashlib

# Crear un diccionario con los IDs reales de los videos
mascara_ids_validos = (datos_seleccionados['video_id'].astype(str).str.strip().str.len() == 11) & \
                      (~datos_seleccionados['video_id'].isin(['#NAME?', '#VALUE!']))

# Mapeamos la combinación title  channela su ID real
diccionario_ids_reales = datos_seleccionados[mascara_ids_validos].set_index(
    ['title', 'channel_title']
)['video_id'].to_dict()

def corregir_identificador_video_pro(fila):
    id_actual = str(fila['video_id']).strip()

    if id_actual in ['#NAME?', '#VALUE!']:
        llave = (fila['title'], fila['channel_title'])
        # AIntentar recuperar el ID real si existió en otra fecha
        if llave in diccionario_ids_reales:
            return diccionario_ids_reales[llave]

        # Si no existe ID real en ninguna fila generamos un Hash único
        texto_combinado = f"{fila['title']}_{fila['channel_title']}"
        return "temp_" + hashlib.md5(texto_combinado.encode('utf-8', errors='ignore')).hexdigest()[:11]

    return id_actual

# Aplicamos la corrección
datos_seleccionados['video_id'] = datos_seleccionados.apply(corregir_identificador_video_pro, axis=1)

restantes_name = (datos_seleccionados['video_id'] == '#NAME?').sum()
print(f"IDs corruptos tipo '#NAME?' restantes: {restantes_name}")
id_nulos = (datos_seleccionados['video_id'].isna()).sum()
print(f"IDs nulos restantes: {id_nulos}")
```

**Eliminar duplicados**

Con los IDs ya corregidos, procedemos a revisar las redundancias técnicas de registro (el mismo video capturado dos o más veces en la misma fecha de tendencia). Para asegurar que el modelo reciba la información más completa y actualizada de vistas acumuladas, ordenamos el conjunto de datos de manera descendente según la variable `views` antes de eliminar las repeticiones, manteniendo así el registro con el mayor número de visualizaciones para cada día.

```python
# Crear una columna auxiliar temporal con valores numéricos de views
# Se elige views y se ordena en forma descendente para obtener siempre el video duplicado con el numero mayor de vistas.

datos_seleccionados['views_temporal'] = pd.to_numeric(datos_seleccionados['views'], errors='coerce')

# Ordenar de forma descendente por vistas para conservar el registro mas alto de cada día
datos_ordenados = datos_seleccionados.sort_values(by='views_temporal', ascending=False)

# Eliminar duplicados basados en video_id y trending_date (Duplicados del mismo dia)
datos_sin_duplicados = datos_ordenados.drop_duplicates(subset=['video_id', 'trending_date'], keep='first').copy()

# Eliminar la columna auxiliar
datos_sin_duplicados = datos_sin_duplicados.drop(columns=['views_temporal'])

duplicados_tecnicos_restantes = datos_sin_duplicados.duplicated(subset=['video_id', 'trending_date']).sum()

reporte_verificacion_descriptivo = pd.DataFrame({
    'Métrica de Control': [
        'Registros antes de de-duplicar',
        'Registros después de de-duplicar',
        'Duplicados del mismo día eliminados',
        'Duplicados del mismo día RESTANTES'
    ],
    'Valor': [
        len(datos_seleccionados),
        len(datos_sin_duplicados),
        len(datos_seleccionados) - len(datos_sin_duplicados),
        duplicados_tecnicos_restantes
    ]
})

print("REPORTE DE VERIFICACIÓN: DATASET DE ANÁLISIS DESCRIPTIVO")
display(reporte_verificacion_descriptivo)
```

**Convertir a los tipos de datos correctos**



A continuación, se detalla el proceso de conversión de tipos de datos para las variables seleccionadas, aplicando estrategias específicas para manejar inconsistencias y valores nulos, y asegurando que cada columna adopte el formato más adecuado para el análisis y el modelado:

```python
datos_sin_duplicados['trending_date'] = pd.to_datetime(datos_sin_duplicados['trending_date'], format='%y.%d.%m')
datos_sin_duplicados['publish_time'] = pd.to_datetime(datos_sin_duplicados['publish_time'], errors='coerce')

columnas_numericas = ['views', 'likes', 'dislikes', 'comment_count', 'category_id']
for col in columnas_numericas:
    datos_sin_duplicados[col] = pd.to_numeric(datos_sin_duplicados[col], errors='coerce').astype('Int64')

columnas_texto = ['title', 'channel_title', 'tags', 'description']
for col in columnas_texto:
    datos_sin_duplicados[col] = datos_sin_duplicados[col].fillna('').astype(str)

mapeo_booleano_a_binario = {'FALSO': 0, 'VERDADERO': 1}
columnas_booleanas = ['comments_disabled', 'ratings_disabled', 'video_error_or_removed']
for col in columnas_booleanas:
    if datos_sin_duplicados[col].dtype == 'object':
        datos_sin_duplicados[col] = datos_sin_duplicados[col].map(mapeo_booleano_a_binario)
    datos_sin_duplicados[col] = datos_sin_duplicados[col].astype('Int64')

print("Tipos de datos convertidos exitosamente.")
```

## Construir nuevos datos

**Creacion de variables derivadas**

Vamos a crear una variable derivada de `trending_date` y `publish_time`, que representara la diferencia de dias entre la fecha de publicacion y la fecha de tendencia.

Con esta variable podemos determinar en cuanto tiempo (en dias) se ha demorado un video en volverse viral.

```python
# dt.tz_localize(None) elimina la zona horaria para tener una diferencia correcta entre ambas fechas, ademas de normalizar la diferencia para comparar solo dias
datos_sin_duplicados['dias_hasta_tendencia'] = (
    datos_sin_duplicados['trending_date'].dt.normalize() -
    datos_sin_duplicados['publish_time'].dt.tz_localize(None).dt.normalize()
).dt.days

# Mostramos estadísticas rápidas de la nueva variable
print("Métricas rápidas de 'dias_hasta_tendencia':")
print(datos_sin_duplicados['dias_hasta_tendencia'].describe())
```

*   **Días hasta tendencia (`dias_hasta_tendencia`):**
    *   La mediana es de **2 días** (la gran mayoría de los videos entra a la lista de tendencias a los dos días de su publicación).
    *   El primer cuartil ($Q1$) es **1 día** y el tercer cuartil ($Q3$) es **3 días**, lo que define un Rango Intercuartil (IQR) de **2 días**.
    *   El límite superior de outliers se establece en **6 días** ($Q3 + 1.5 \times IQR$).
    *   Cualquier video que tarde **más de 6 días** en entrar a tendencias (llegando a registrar el valor máximo de **221 días**) es considerado un **outlier de viralización tardía**.

Para responder de forma directa algunas preguntas como la 3 y 4, se deciden crear estas variables:

- **ratio_likes_dislikes:** (Proporcion Me gusta/No me gusta)
- **ratio_vistas_comentarios:** (Proporcion Vistas/Comentarios)

```python
datos_sin_duplicados['ratio_likes_dislikes'] = (
    datos_sin_duplicados['likes'] / (datos_sin_duplicados['dislikes'] + 1)
)

datos_sin_duplicados['ratio_vistas_comentarios'] = (
    datos_sin_duplicados['views'] / (datos_sin_duplicados['comment_count'] + 1)
)

print("Variables de ratio creadas exitosamente.")
display(datos_sin_duplicados[['ratio_likes_dislikes', 'ratio_vistas_comentarios']].describe())
```

**Deteccion de valores atipicos (numericos)**





```python

sns.set_theme(style="whitegrid")

# Variables de interaccion
fig, ejes = plt.subplots(4, 2, figsize=(16, 20))
fig.suptitle('Distribución y Detección Visual de Outliers en Métricas de Interacción', fontsize=18, fontweight='bold')

columnas_metricas = ['views', 'likes', 'dislikes', 'comment_count']
titulos_metricas = ['Vistas', 'Me Gusta', 'No Me Gusta', 'Comentarios']

for idx, col in enumerate(columnas_metricas):
    datos_grafico = datos_sin_duplicados[col].dropna().astype(float)

    sns.boxplot(ax=ejes[idx, 0], x=datos_grafico, color="skyblue")
    ejes[idx, 0].set_title(f'Diagrama de Caja - {titulos_metricas[idx]}', fontsize=12, fontweight='bold')
    ejes[idx, 0].set_xlabel('')

    sns.histplot(ax=ejes[idx, 1], x=datos_grafico, bins=50, kde=True, color="skyblue")
    ejes[idx, 1].set_title(f'Histograma - {titulos_metricas[idx]}', fontsize=12, fontweight='bold')
    ejes[idx, 1].set_xlabel('')
    ejes[idx, 1].set_ylabel('Frecuencia')

plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()

# Variables geograficas
fig, ejes_geo = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('Distribución y Detección Visual de Outliers en Coordenadas Geográficas', fontsize=16, fontweight='bold')

columnas_geo = ['lat', 'lon']
titulos_geo = ['Latitud (lat)', 'Longitud (lon)']

for idx, col in enumerate(columnas_geo):
    datos_grafico = datos_sin_duplicados[col].dropna()

    # Boxplot
    sns.boxplot(ax=ejes_geo[idx, 0], x=datos_grafico, color="lightgreen")
    ejes_geo[idx, 0].set_title(f'Diagrama de Caja - {titulos_geo[idx]}', fontsize=12, fontweight='bold')
    ejes_geo[idx, 0].set_xlabel('')
    # Histograma
    sns.histplot(ax=ejes_geo[idx, 1], x=datos_grafico, bins=30, kde=True, color="lightgreen")
    ejes_geo[idx, 1].set_title(f'Histograma - {titulos_geo[idx]}', fontsize=12, fontweight='bold')
    ejes_geo[idx, 1].set_xlabel('')
    ejes_geo[idx, 1].set_ylabel('Frecuencia')

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

print("CALCULO DE LIMITES E IDENTIFICACION DE OUTLIERS (IQR)\n")
for col in columnas_metricas + columnas_geo:
    datos_columna = datos_sin_duplicados[col].dropna().astype(float)
    q1 = datos_columna.quantile(0.25)
    q3 = datos_columna.quantile(0.75)
    iqr = q3 - q1
    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    atipicos = datos_columna[(datos_columna < limite_inferior) | (datos_columna > limite_superior)]
    porcentaje_atipicos = (len(atipicos) / len(datos_columna)) * 100

    print(f"Variable: {col}")
    print(f"  Rango Intercuartil (IQR): {iqr:.2f} (Q1: {q1:.2f} | Q3: {q3:.2f})")
    print(f"  Límites Estadisticos: [{limite_inferior:.2f} a {limite_superior:.2f}]")
    print(f"  Valor Maximo Registrado: {datos_columna.max():.2f}")
    print(f"  Outliers Detectados: {len(atipicos)} ({porcentaje_atipicos:.2f}%)")
    print("-" * 50)
```

Debido a que es físicamente imposible tener interacciones negativas, el límite inferior real es **0**, lo que significa que no existen outliers en la parte baja de la distribución.
    *   **Justificación de Tratamiento:** Todos estos valores extremos superiores (con máximos de hasta 125M de vistas y 2.9M de me gusta) corresponden a videos virales legítimos.
```python
# calcular los cuartiles y el Rango Intercuartil
tiempo_viralizacion = datos_sin_duplicados['dias_hasta_tendencia'].dropna().astype(float)
q1 = tiempo_viralizacion.quantile(0.25)
q3 = tiempo_viralizacion.quantile(0.75)
iqr = q3 - q1
limite_superior_outliers = q3 + 1.5 * iqr

# identificar y contar los registros atipicos
atipicos_tiempo = tiempo_viralizacion[tiempo_viralizacion > limite_superior_outliers]
porcentaje_atipicos_tiempo = (len(atipicos_tiempo) / len(tiempo_viralizacion)) * 100

print("REPORTE DE OUTLIERS: DÍAS HASTA TENDENCIA")
print(f"Cuartil 1 (25%): {q1:.1f} dias")
print(f"Mediana (50%): {tiempo_viralizacion.median():.1f} dias")
print(f"Cuartil 3 (75%): {q3:.1f} dias")
print(f"Rango Intercuartil (IQR): {iqr:.1f} dias")
print(f"Limite Superior Estadistico: {limite_superior_outliers:.1f} dias")
print(f"Persistencia maxima en Datos: {tiempo_viralizacion.max():.1f} dias")
print(f"Cantidad de Outliers: {len(atipicos_tiempo)} registros ({porcentaje_atipicos_tiempo:.2f}%)")

# graficar el diagrama de caja para la visualización del sesgo
plt.figure(figsize=(10, 4))
sns.boxplot(x=tiempo_viralizacion, color="salmon")
plt.title('Detección Visual de Outliers en Días Transcurridos hasta Tendencias', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Días transcurridos (Publicación a Tendencia)', fontsize=12)
plt.tight_layout()
plt.show()
```

**Tratamiento de Outliers y Transformación de Datos**

El preprocesamiento de los valores atípicos se define según el objetivo analítico de cada fase del proyecto, diferenciando el análisis descriptivo del modelado de regresión.

---

**1. Tratamiento de Valores Atípicos (Outliers)**

**A. Tratamiento para el Análisis Descriptivo (Preguntas 1 a 8)**
*   **Estrategia:** **No se modificará ni eliminará ningún outlier.**
*   **Justificación:** Los videos virales extremos representan el núcleo del negocio y la realidad de la plataforma. Suprimir estos datos subestimaría la influencia de las categorías y canales líderes en el reporte descriptivo para el cliente.

**B. Tratamiento para el Modelado Predictivo (Regresión Lineal - Pregunta 9)**
Para evitar que las observaciones extremas desestabilicen el modelo de regresión lineal y sesguen los coeficientes, se aplicarán los siguientes tratamientos antes del entrenamiento:
*   **Winsorización al Percentil 99 ($P_{99}$):** Se aplicará esta técnica sobre las variables métricas del modelo (`likes`, `views`, `dislikes` y `comment_count`). Aquellos registros que superen el percentil 99 serán limitados al valor de dicho percentil (por ejemplo, la variable objetivo `likes` se topará en un máximo de 402,776). Esto reduce la influencia de los videos hiper-virales sin necesidad de descartar observaciones valiosas.
*   **Exclusión de Anomalías y Restricciones:** Se filtrarán y excluirán del conjunto de datos los videos con errores de carga (`video_error_or_removed == 1`) y los videos con calificaciones desactivadas por el creador (`ratings_disabled == 1`), garantizando que el modelo lineal aprenda únicamente del comportamiento de interacción real y orgánica de la audiencia.

---

**2. Variables del Modelo: Tratamiento y Transformación**

A continuación, se detalla la lista de variables seleccionadas para el modelo de **Regresión Lineal Múltiple**, especificando el tratamiento de sus outliers y su posterior método de transformación:

1.  **`likes` (Variable Objetivo / Target):**
    *   **Tratamiento de Outliers:** Sí. Winsorización al percentil 99 ($P_{99}$).
2.  **`views` (Predictor cuantitativo):**
    *   **Tratamiento de Outliers:** Sí. Winsorización al percentil 99 ($P_{99}$).
3.  **`dislikes` (Predictor cuantitativo):**
    *   **Tratamiento de Outliers:** Sí. Winsorización al percentil 99 ($P_{99}$).
4.  **`comment_count` (Predictor cuantitativo):**
    *   **Tratamiento de Outliers:** Sí. Winsorización al percentil 99 ($P_{99}$).
5.  **`nombre_categoria` (Predictor cualitativo):**
    *   **Tratamiento de Outliers:** No (las categorías minoritarias se conservan).
    *   **Transformación de Datos:** Generalización mediante Variables Dummy.

*(Nota: Las variables `comments_disabled`, `ratings_disabled`, `video_error_or_removed`, `dias_hasta_tendencia`, `lat` y `lon` fueron excluidas del modelado por justificación estadística y empírica, evitando la redundancia de datos y las colinealidades constantes en el dataset de modelado).*

Exclusión de Videos con Errores (`video_error_or_removed`)

```python
datos_modelo = datos_sin_duplicados.copy()
filas_previas = len(datos_modelo)

datos_modelo = datos_modelo[datos_modelo['video_error_or_removed'] == 0]
datos_modelo = datos_modelo[datos_modelo['ratings_disabled'] == 0]

columnas_modelo = ['video_id', 'trending_date', 'likes', 'views', 'dislikes', 'comment_count', 'nombre_categoria']
datos_modelo = datos_modelo[columnas_modelo]

filas_posteriores = len(datos_modelo)

print("PREPARACIÓN DE DATOS PARA EL MODELO PREDICTIVO")
print(f"Filas iniciales (Dataset Descriptivo): {filas_previas}")
print(f"Filas resultantes en 'datos_modelo': {filas_posteriores}")
print(f"Filas eliminadas por filtros: {filas_previas - filas_posteriores}")
print(f"\nColumnas conservadas en 'datos_modelo': {list(datos_modelo.columns)}")
```

Se identificaron y eliminaron del dataset de modelado los videos con errores de carga o dados de baja de la plataforma, así como aquellos videos cuyas calificaciones de me gusta/no me gusta fueron desactivadas por el creador (710 registros). Dado que un video inactivo o con calificaciones bloqueadas no aporta información orgánica sobre la propensión a interactuar, su exclusión es necesaria para evitar sesgos numéricos. El volumen del dataset se redujo a 31,990 observaciones listas para el entrenamiento.

Tratamiento de Outliers en `views`

```python
limite_views_p99 = datos_modelo['views'].quantile(0.99)

# Crear una copia de la columna original para poder graficar la comparación
views_original = datos_modelo['views'].astype(float).copy()

# Aplicar Winsorización: reemplazar cualquier valor por encima del P99
datos_modelo['views'] = np.where(datos_modelo['views'] > limite_views_p99, limite_views_p99, datos_modelo['views'])

# Graficar la comparación de antes" vs Despues
fig, ejes = plt.subplots(1, 2, figsize=(15, 4))
fig.suptitle('Comparación de Distribución de Vistas: Antes vs. Después del Tratamiento', fontsize=14, fontweight='bold')

# Antes
sns.boxplot(ax=ejes[0], x=views_original, color="orange")
ejes[0].set_title('views (Antes del Tratamiento - Con Outliers)', fontsize=11, fontweight='bold')
ejes[0].set_xlabel('Número de Vistas')

# Después
sns.boxplot(ax=ejes[1], x=datos_modelo['views'].astype(float), color="skyblue")
ejes[1].set_title(f'views (Después del Tratamiento - Capped al P99: {limite_views_p99:,.0f})', fontsize=11, fontweight='bold')
ejes[1].set_xlabel('Número de Vistas (Escala Capped)')

plt.tight_layout()
plt.show()

print(f"Límite del Percentil 99: {limite_views_p99:,.2f} vistas.")
print(f"Valor máximo original: {views_original.max():,.2f} vistas.")
print(f"Nuevo valor máximo tratado: {datos_modelo['views'].max():,.2f} vistas.")
```

Aplicar la Winsorización en lugar de eliminar los outliers o imputarlos por la media es la estrategia correcta para el modelado de regresión lineal. Eliminar los atípicos descartaría entre el 11% y 15% del dataset, incluyendo todos los videos virales (más de 1.6M de vistas), lo que impediría al modelo predecir el éxito de videos populares. Por otro lado, reemplazarlos por el promedio rompería la relación lógica entre las variables, generando inconsistencias matemáticas insostenibles para la regresión (como videos con más "likes" que visualizaciones).

La Winsorización al percentil 99 de las vistas resuelve estos problemas al conservar la totalidad de los datos y sus proporciones relativas, pero acotando los valores extremos. Al topar el máximo de vistas en 12.1 millones (reduciendo la escala original en 10 veces), se neutraliza el efecto de "puntos de apalancamiento" que desestabilizan el algoritmo de regresión lineal. De esta manera, las vistas actúan como un predictor estable de la variable objetivo (likes) sin verse afectado por valores hiper-virales aislados.

Tratamiento de Outliers en `likes`

```python
limite_likes_p99 = datos_modelo['likes'].quantile(0.99)

likes_original = datos_modelo['likes'].astype(float).copy()

# reemplazar cualquier valor por encima del P99
datos_modelo['likes'] = np.where(datos_modelo['likes'] > limite_likes_p99, limite_likes_p99, datos_modelo['likes'])

# graficar la comparación de Antes vs Después
fig, ejes = plt.subplots(1, 2, figsize=(15, 4))
fig.suptitle('Comparación de Distribución de Me Gusta: Antes vs. Después del Tratamiento', fontsize=14, fontweight='bold')

# Antes
sns.boxplot(ax=ejes[0], x=likes_original, color="orange")
ejes[0].set_title('likes (Antes del Tratamiento - Con Outliers)', fontsize=11, fontweight='bold')
ejes[0].set_xlabel('Cantidad de Me Gusta')

# Después
sns.boxplot(ax=ejes[1], x=datos_modelo['likes'].astype(float), color="skyblue")
ejes[1].set_title(f'likes (Después del Tratamiento - Capped al P99: {limite_likes_p99:,.0f})', fontsize=11, fontweight='bold')
ejes[1].set_xlabel('Cantidad de Me Gusta (Escala Capped)')

plt.tight_layout()
plt.show()

print(f"Límite del Percentil 99: {limite_likes_p99:,.2f} me gusta.")
print(f"Valor máximo original: {likes_original.max():,.2f} me gusta.")
print(f"Nuevo valor máximo tratado: {datos_modelo['likes'].max():,.2f} me gusta.")
```

Se aplicó la Winsorización al percentil 99 ($P_{99}$) sobre la variable objetivo `likes`, limitando su valor máximo de `2,912,710` a `402,776`. Al ser `likes` la variable dependiente de nuestra regresión lineal múltiple, topar los valores atípicos mediante winsorización previene que la estimación del error cuadrático del modelo se vea desproporcionadamente influenciada por contados videos con reacciones positivas extraordinarias, mejorando la precisión y la generalización del pronóstico global.

Tratamiento de Outliers en `dislikes`

```python
# percentil 99 de dislikes
limite_dislikes_p99 = datos_modelo['dislikes'].quantile(0.99)

dislikes_original = datos_modelo['dislikes'].astype(float).copy()

# reemplazar cualquier valor por encima del P99
datos_modelo['dislikes'] = np.where(datos_modelo['dislikes'] > limite_dislikes_p99, limite_dislikes_p99, datos_modelo['dislikes'])

fig, ejes = plt.subplots(1, 2, figsize=(15, 4))
fig.suptitle('Comparación de Distribución de No Me Gusta: Antes vs. Después del Tratamiento', fontsize=14, fontweight='bold')

# Antes
sns.boxplot(ax=ejes[0], x=dislikes_original, color="orange")
ejes[0].set_title('dislikes (Antes del Tratamiento - Con Outliers)', fontsize=11, fontweight='bold')
ejes[0].set_xlabel('Cantidad de No Me Gusta')

# Después
sns.boxplot(ax=ejes[1], x=datos_modelo['dislikes'].astype(float), color="skyblue")
ejes[1].set_title(f'dislikes (Después del Tratamiento - Capped al P99: {limite_dislikes_p99:,.0f})', fontsize=11, fontweight='bold')
ejes[1].set_xlabel('Cantidad de No Me Gusta (Escala Capped)')

plt.tight_layout()
plt.show()

print(f"Límite del Percentil 99: {limite_dislikes_p99:,.2f} no me gusta.")
print(f"Valor máximo original: {dislikes_original.max():,.2f} no me gusta.")
print(f"Nuevo valor máximo tratado: {datos_modelo['dislikes'].max():,.2f} no me gusta.")
```

El valor máximo de reacciones negativas se limitó de `1,545,017` a `19,170`, comprimiendo significativamente la dispersión extrema de la variable. Controlar esta escala es crítico para el modelo de regresión lineal, ya que los videos con controversias masivas (outliers de dislikes) suelen actuar como fuertes anomalías numéricas. Al acotar este rango al percentil 99, estabilizamos la varianza y evitamos coeficientes sesgados por videos de polémica atípica.

Analizaremos la suma total de interacciones agrupadas por la variable `state`. Utilizaremos una visualización de múltiples facetas para comparar el alcance (vistas) frente a la reacción (likes/dislikes).

```python
# Agrupamos las métricas por Estado
metricas_por_estado = datos_sin_duplicados.groupby('state')[['views', 'likes', 'dislikes']].sum().reset_index()

# Ordenamos los estados por volumen de vistas para una lectura jerárquica
orden_estados = metricas_por_estado.sort_values(by='views', ascending=False)['state']

# Identificamos el Top 3 de estados con más vistas para el resumen
top_estados = metricas_por_estado.nlargest(3, 'views')
display(top_estados[['state', 'views', 'likes', 'dislikes']].style.format({'views': '{:,}', 'likes': '{:,}', 'dislikes': '{:,}'}))
```

```python

import plotly.express as px

# agrupamos por estado
reporte_estados = datos_sin_duplicados.groupby('state').agg({
    'views': 'sum',
    'likes': 'sum',
    'dislikes': 'sum',
    'lat': 'first',
    'lon': 'first'
}).reset_index()

# identificamos el estado para cada metrica
estado_mas_vistas = reporte_estados.loc[reporte_estados['views'].idxmax(), 'state']
estado_mas_likes = reporte_estados.loc[reporte_estados['likes'].idxmax(), 'state']
estado_mas_dislikes = reporte_estados.loc[reporte_estados['dislikes'].idxmax(), 'state']

# columna descriptiva para el del mapa
def clasificar_liderazgo(fila):
    logros = []
    if fila['state'] == estado_mas_vistas:
        logros.append("Más Vistas")
    if fila['state'] == estado_mas_likes:
        logros.append("Más Likes")
    if fila['state'] == estado_mas_dislikes:
        logros.append("Más Dislikes")
    return " / ".join(logros)

reporte_estados['Líder en'] = reporte_estados.apply(clasificar_liderazgo, axis=1)

# filtramos el DataFrame para conservar solo a los estos
estados_top = [estado_mas_vistas, estado_mas_likes, estado_mas_dislikes]
reporte_filtrado = reporte_estados[reporte_estados['state'].isin(estados_top)].copy()

# convertimos a float para Plotly
for col in ['views', 'likes', 'dislikes']:
    reporte_filtrado[col] = reporte_filtrado[col].astype(float)

# graficamos
fig = px.scatter_mapbox(
    reporte_filtrado,
    lat='lat',
    lon='lon',
    color='Líder en',
    size='views',
    size_max=20,
    hover_name='state',
    hover_data={
        'lat': ':.4f',
        'lon': ':.4f',
        'views': ':.0f',
        'likes': ':.0f',
        'dislikes': ':.0f',
        'Líder en': True
    },
    zoom=3.8,
    center=dict(lat=22.0, lon=78.0),
    title='<b>Geografía de Tendencias: Estados Líderes en Vistas, Likes y Dislikes en la India</b>'
)

# Configurar el estilo del mapa
fig.update_layout(
    mapbox_style="open-street-map",
    margin={"r":0,"t":50,"l":0,"b":0},
    height=600
)

# Mostrar mapa
fig.show()
```

El análisis geográfico del confirma la homogeneidad del consumo de tendencias en la India. A pesar de la vasta diversidad cultural del país, el volumen de visualizaciones y reacciones se mantiene sorprendentemente altamente equitativo entre todos los estados analizados.

## Construccion del modelo

La fase de construcción del modelo inicia con la separación de los predictores en variables cuantitativas y categóricas para los subconjuntos de entrenamiento y prueba. Acto seguido, se procede a la estandarización de los datos numéricos mediante el uso de la técnica StandardScaler, con el objetivo de situar todas las variables en una escala común y evitar sesgos por magnitudes distintas. Este procedimiento incluye el ajuste del escalador exclusivamente con la información de entrenamiento para el cálculo de los parámetros estadísticos. Finalmente, se aplica la transformación tanto al conjunto de entrenamiento como al de prueba, lo cual garantiza la homogeneidad de los datos y asegura la integridad del proceso de validación antes de la ejecución del algoritmo de regresión.

**Inicializar y ajustar el escalador únicamente con los datos de entrenamiento**

```python
escalador = StandardScaler()
```

Creamos DataFrames con los datos ya escalados manteniendo el mismo índice original

```python
X_train_num_scaled = pd.DataFrame(
    escalador.fit_transform(X_train_num),
    columns=columnas_cuantitativas,
    index=X_train.index
)
```

**Transformamos los datos de prueba usando el mismo escalador (transform)**

```python
X_test_num_scaled = pd.DataFrame(
    escalador.transform(X_test_num),
    columns=columnas_cuantitativas,
    index=X_test.index
)
```

**Fusionar (merge) las variables escaladas con las dummies usando left_index y right_index**

```python
X_train_preprocessed = pd.merge(X_train_num_scaled, X_train_cat, left_index=True, right_index=True)
X_test_preprocessed = pd.merge(X_test_num_scaled, X_test_cat, left_index=True, right_index=True)
```

**Mostrar los datos procesados**

```python
print("Datos de Entrenamiento Preprocesados (Vista Previa):")
display(X_train_preprocessed.head())
```

**Entrenamiento del modelo**

```python
modelo_regresion = LinearRegression()
modelo_regresion.fit(X_train_preprocessed, y_train)
```

**Generar las predicciones del conjunto de prueba**

```python
predicciones_reales = modelo_regresion.predict(X_test_preprocessed)
y_test_real = y_test.values
```

## Evaluacion  del modelo

La evaluación del rendimiento del modelo se sustenta en el MAE, MSE y RMSE, métricas que cuantifican la magnitud del error en las predicciones. El MAE proporciona una medición directa en unidades de «me gusta», mientras que el RMSE penaliza desviaciones mayores, resultando crítico para detectar fallos en videos de alta viralidad. La comparación de estos indicadores frente a la escala real de los datos valida la capacidad predictiva del algoritmo, confirmando que el modelo funciona como una herramienta robusta para la estimación de interacciones y la optimización de estrategias publicitarias.


**Calculo de metricas de evaluacion**

**MAE (Error Absoluto Medio)**

```python
mae = metrics.mean_absolute_error(y_test_real, predicciones_reales)
```

**MSE (Error Cuadrático Medio)**

```python
mse = metrics.mean_squared_error(y_test_real, predicciones_reales)
```

**RMSE (Raíz del Error Cuadrático Medio)**

```python
rmse = np.sqrt(mse)
```
